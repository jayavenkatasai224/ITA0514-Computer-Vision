```python

import cv2
from google.colab import files
from google.colab.patches import cv2_imshow

# Upload flower image
uploaded = files.upload()

# Read uploaded image
filename = next(iter(uploaded))
img = cv2.imread(filename)

if img is None:
    print("Image not found")
else:
    # Get image dimensions
    height, width = img.shape[:2]

    # Define Region of Interest (ROI)
    x1 = width // 4
    y1 = height // 4
    x2 = 3 * width // 4
    y2 = 3 * height // 4

    # Crop the selected region
    roi = img[y1:y2, x1:x2]

    # Copy the ROI
    copied_roi = roi.copy()

    # Create a copy of original image
    result = img.copy()

    # Paste ROI into another region
    h, w = copied_roi.shape[:2]
    result[0:h, 0:w] = copied_roi

    # Display images
    print("Original Image")
    cv2_imshow(img)

    print("Cropped ROI")
    cv2_imshow(roi)

    print("Copied ROI")
    cv2_imshow(copied_roi)

    print("Image After Pasting ROI")
    cv2_imshow(result)
```
