```python

import cv2
from google.colab import files
from google.colab.patches import cv2_imshow

# Upload flower image
uploaded = files.upload()

# Read the image
img = cv2.imread(next(iter(uploaded)))

if img is None:
    print("Image not found")
else:
    # Get original dimensions
    height, width = img.shape[:2]

    print("Original Image Size:", width, "x", height)

    # Scale image to 50%
    scaled_down = cv2.resize(
        img, None, fx=0.5, fy=0.5,
        interpolation=cv2.INTER_LINEAR
    )

    # Scale image to 150%
    scaled_up = cv2.resize(
        img, None, fx=1.5, fy=1.5,
        interpolation=cv2.INTER_LINEAR
    )

    # Display images
    print("Original Image")
    cv2_imshow(img)

    print("Scaled Down Image (50%)")
    cv2_imshow(scaled_down)

    print("Scaled Up Image (150%)")
    cv2_imshow(scaled_up)

    # Save output images
    cv2.imwrite("scaled_down.jpg", scaled_down)
    cv2.imwrite("scaled_up.jpg", scaled_up)

    # Download output
    files.download("scaled_down.jpg")
    files.download("scaled_up.jpg")
```
