```python

import cv2
from google.colab import files
from IPython.display import Video, display

# Upload video
uploaded = files.upload()

# Get uploaded video filename
video_path = next(iter(uploaded))

# Open video
cap = cv2.VideoCapture(video_path)

if not cap.isOpened():
    print("Video not found")
else:
    # Get video properties
    fps = cap.get(cv2.CAP_PROP_FPS)
    width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

    # Read all frames
    frames = []

    while True:
        ret, frame = cap.read()
        if not ret:
            break
        frames.append(frame)

    cap.release()

    if len(frames) == 0:
        print("No frames found")
    else:
        fourcc = cv2.VideoWriter_fourcc(*'mp4v')

        # Create slow motion video
        slow = cv2.VideoWriter(
            "slow_motion.mp4", fourcc,
            fps / 2, (width, height)
        )

        for frame in frames:
            slow.write(frame)

        slow.release()

        # Create fast motion video
        fast = cv2.VideoWriter(
            "fast_motion.mp4", fourcc,
            fps * 2, (width, height)
        )

        for i in range(0, len(frames), 2):
            fast.write(frames[i])

        fast.release()

        print("Slow Motion Video")
        display(Video("slow_motion.mp4", embed=True))

        print("Fast Motion Video")
        display(Video("fast_motion.mp4", embed=True))

        # Download videos
        files.download("slow_motion.mp4")
        files.download("fast_motion.mp4")
```
