```python

import cv2
import numpy as np
from google.colab.patches import cv2_imshow

# Create a white image
img = np.ones((500, 700, 3), dtype=np.uint8) * 255

# Get text from the user
text = input("Enter the text: ")

# Display text on the image
cv2.putText(
    img,
    text,
    (50, 250),
    cv2.FONT_HERSHEY_SIMPLEX,
    1.5,
    (0, 0, 255),
    2,
    cv2.LINE_AA
)

# Display the image
cv2_imshow(img)
```
