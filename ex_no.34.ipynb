```python

import cv2
import numpy as np
from google.colab.patches import cv2_imshow

# Create a white image
img = np.ones((500, 500, 3), dtype=np.uint8) * 255

# Draw a circle
center = (250, 250)
radius = 100

# Circle color: Red
color = (0, 0, 255)

# Draw circle
cv2.circle(img, center, radius, color, 3)

# Display image
cv2_imshow(img)
```
